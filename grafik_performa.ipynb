```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# 1. Baca data dari file CSV yang sudah Anda buat sebelumnya
df = pd.read_csv('laporan_performa_flappybird.csv')

# 2. Cari Checkpoint Terbaik secara otomatis
best_row = df.loc[df['Rata-rata Skor'].idxmax()]
best_iter = int(best_row['Iterasi Checkpoint'])
best_avg = best_row['Rata-rata Skor']

# 3. Setup gaya visual agar terlihat profesional dan modern (ala presentasi akademik)
sns.set_theme(style="darkgrid")
plt.figure(figsize=(14, 7))

# 4. Gambar rentang variansi (Skor tertinggi ke terendah) dengan warna transparan
plt.fill_between(df['Iterasi Checkpoint'], df['Skor Terendah'], df['Skor Tertinggi'], 
                 alpha=0.2, color='royalblue', label='Rentang Skor (Min - Max)')

# 5. Gambar garis Rata-rata Skor
sns.lineplot(x='Iterasi Checkpoint', y='Rata-rata Skor', data=df, 
             linewidth=3, color='navy', label='Rata-rata Skor')

# 6. Beri tanda bintang pada titik performa terbaik
plt.scatter(best_iter, best_avg, color='red', s=200, zorder=5, marker='*', 
            label=f'Model Terbaik ({best_iter/1000:.0f}k)')

# 7. Tambahkan teks penjelasan di titik terbaik
plt.text(best_iter, best_avg + 15, f' Avg: {best_avg:.1f}', color='darkred', 
         fontweight='bold', horizontalalignment='center')

# 8. Kostumisasi label dan judul
plt.title('Kurva Pembelajaran Deep Q-Network pada Flappy Bird', fontsize=18, fontweight='bold', pad=20)
plt.xlabel('Iterasi Pelatihan', fontsize=14)
plt.ylabel('Skor Rintangan Pipa', fontsize=14)
plt.xticks(np.arange(0, 1100000, 100000), [f'{i//1000}k' for i in range(0, 1100000, 100000)])
plt.legend(fontsize=12, loc='upper left')

# 9. Simpan grafik sebagai gambar PNG beresolusi tinggi (siap masuk presentasi/PDF)
plt.tight_layout()
plt.savefig('Visualisasi_Kurva_Belajar.png', dpi=300)
plt.show()

print("="*50)
print(f"✅ Gambar grafik berhasil disimpan: Visualisasi_Kurva_Belajar.png")
print(f"🏆 PENTING: Pindahkan file 'checkpoint_{best_iter}.pt' ke folder utama untuk dosen!")
print("="*50)
```
